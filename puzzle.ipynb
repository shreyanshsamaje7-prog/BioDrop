```python
# Dynamic 8-Puzzle using Iterative Deepening Search (IDS)

def print_board(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


def get_next_states(state):
    """Generate all valid moves of the blank (0)."""

    next_states = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    # UP
    if row > 0:
        new_state = list(state)
        new_state[blank], new_state[blank - 3] = \
            new_state[blank - 3], new_state[blank]
        next_states.append(("UP", tuple(new_state)))

    # DOWN
    if row < 2:
        new_state = list(state)
        new_state[blank], new_state[blank + 3] = \
            new_state[blank + 3], new_state[blank]
        next_states.append(("DOWN", tuple(new_state)))

    # LEFT
    if col > 0:
        new_state = list(state)
        new_state[blank], new_state[blank - 1] = \
            new_state[blank - 1], new_state[blank]
        next_states.append(("LEFT", tuple(new_state)))

    # RIGHT
    if col < 2:
        new_state = list(state)
        new_state[blank], new_state[blank + 1] = \
            new_state[blank + 1], new_state[blank]
        next_states.append(("RIGHT", tuple(new_state)))

    return next_states


def depth_limited_search(state, goal, limit, path, visited):

    # 1. Goal test
    if state == goal:
        return path

    # 2. Depth limit reached
    if limit == 0:
        return None

    # 3. Generate possible moves
    for move, next_state in get_next_states(state):

        # 4. Avoid states already in current path
        if next_state not in visited:

            visited.add(next_state)

            result = depth_limited_search(
                next_state,
                goal,
                limit - 1,
                path + [move],
                visited
            )

            # If solution found
            if result is not None:
                return result

            # Backtrack
            visited.remove(next_state)

    # 5. Failure
    return None


def IDS(start, goal):

    depth = 0

    while True:

        print("\nSearching at depth:", depth)

        visited = {start}

        result = depth_limited_search(
            start,
            goal,
            depth,
            [],
            visited
        )

        if result is not None:
            return result

        depth += 1


# ------------------------------------------------
# MAIN PROGRAM
# ------------------------------------------------

print("8-PUZZLE USING ITERATIVE DEEPENING SEARCH")
print("------------------------------------------")

print("\nEnter the initial state.")
print("Use 0 for the blank space.")
print("Example: 1 2 3 4 0 6 7 5 8")

start = tuple(map(int, input("Initial state: ").split()))

print("\nEnter the goal state.")
print("Example: 1 2 3 4 5 6 7 8 0")

goal = tuple(map(int, input("Goal state: ").split()))


# Validate input
if len(start) != 9 or len(goal) != 9:
    print("Error: Enter exactly 9 numbers.")
    exit()

if set(start) != set(range(9)) or set(goal) != set(range(9)):
    print("Error: Use numbers 0 to 8 exactly once.")
    exit()


print("\nInitial State:")
print_board(start)

print("Goal State:")
print_board(goal)


# Solve
solution = IDS(start, goal)


# Display solution
print("\nSolution Found!")
print("Moves:", " -> ".join(solution))
print("Total moves:", len(solution))


# Display each step
print("\nSolution Path:")
print("Initial:")
print_board(start)

current = start

for i, move in enumerate(solution, 1):

    for m, next_state in get_next_states(current):

        if m == move:
            current = next_state
            break

    print("Move", i, ":", move)
    print_board(current)
```
